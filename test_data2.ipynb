```python
def plot_arima_graphs(data,col,time_period):
    #plot series of respective time series

    data = pd.DataFrame({
        'datetime': data[time_period],
        'series': data[col]
    })

    data = data.set_index('datetime')

    fig1, ax1 = plt.subplots()

    fig13 = ax1.plot(data,color='red',label='data')
    
    #ar,ma,armax
    fig2,ax2 = plt.subplots()
    fig23 = plot_acf(data,lags=20,alpha=0.05,ax=ax2)
    

    fig3,ax3 = plt.subplots()
    fig33 = plot_pacf(data,lags=20,alpha=0.05,ax=ax3)

    return fig13,fig23,fig33
```
